```python
print("Rows before key-based duplicate removal:",len(df))
df["_missing"] = df.isnull().sum(axis=1)
df = (df.sort_values("_missing")
        .drop_duplicates(subset=["Order_ID", "Product_ID"], keep="first")
        .drop(columns="_missing")
        .sort_index())
print("Rows after key-based duplicate removal:",len(df))
```

```python
#create the flag for missing ship_date which indicates which dates are calculated
df['Ship_Date_Estimated']=df['Ship_Date'].isna()
# Filling missing values in Product_Name,Category,Unit_Price
for col in ["Product_Name","Category","Unit_Price"]:
    df[col]=df[col].fillna(df.groupby("Product_ID")[col].transform("first"))
```

```python
df['Region']=df['Region'].fillna(df.groupby('City')['Region'].transform('first'))
df['Region']=df['Region'].fillna("Unknown")
df['City']=df['City'].fillna("Unknown")
```

```python
for col in ['Customer_Name','Segment']:
    df[col]=df[col].fillna(df.groupby('Customer_ID')[col].transform("first"))
df['Customer_Name']=df['Customer_Name'].fillna("Unknown")
df['Customer_ID']=df['Customer_ID'].fillna("Unknown")
df['Segment']=df['Segment'].fillna(df['Segment'].mode()[0])
df['Ship_Mode']=df['Ship_Mode'].fillna(df['Ship_Mode'].mode()[0])
```

```python
allowed = np.array([0,0.05,0.10,0.15,0.20,0.30]) 
qty_f = df['Quantity'].astype('float64')
implied = 1 - df['Sales'] / (df['Unit_Price'] * qty_f) 
snapped = pd.Series(
    allowed[np.abs(implied.fillna(0).to_numpy()[:, None] - allowed).argmin(axis=1)],
    index=df.index) 
mask = df['Discount'].isna() & implied.notna()
df.loc[mask, 'Discount'] = snapped[mask]

df['Discount']=df['Discount'].fillna(df['Discount'].mode()[0])
derived=(df['Sales']/(df['Unit_Price'] * (1-df['Discount']))).round()
df['Quantity']=df['Quantity'].fillna(derived)
df['Quantity']=df['Quantity'].fillna(df['Quantity'].median())
```

```python
# filling the sales
calSales=df['Unit_Price'] * df['Quantity'] * (1-df['Discount'])
df['Sales']=df['Sales'].fillna(calSales.round(2))
#filling the profit
df['Profit']=df['Profit'].fillna(df.groupby('Category')['Profit'].transform("median"))

```

```python
# filling the ship_date
df['Ship_Date']=df['Ship_Date'].fillna(df['Order_Date']+pd.Timedelta(days=4))
df['Quantity']=df['Quantity'].astype('Int64')
```

```python
df=pd.read_csv(f"{CLEANED_DIR}/Cleaned_Retail_Sales.csv")
df['Order_Date']=pd.to_datetime(df['Order_Date'])
df['Ship_Date']=pd.to_datetime(df['Ship_Date'])
```

```python
# Create the Date Features
df['Year']=df['Order_Date'].dt.year
df['Month']=df['Order_Date'].dt.month
df['Month_Name']=df['Order_Date'].dt.month_name()
df['Quarter']=df['Order_Date'].dt.quarter
```

```python
# Profit Margin
df['Profit_Margin']=((df['Profit']/df['Sales'])*100).round(2)

# Margin Outlier Flag
df['Margin_Outlier']=np.where(df['Profit_Margin']>100,'Yes','No')

# Order Value of one row before discount
df['Gross_Line_Value']=(df['Unit_Price'] * df['Quantity']).round(2)

# Total value of the whole order (all items sharing an Order_ID)
df["Order_Value"] = df.groupby("Order_ID")["Sales"].transform("sum").round(2)

```

```python
# Customer Segmentation
known=df["Customer_ID"] != "Unknown" 
customer_spend=df[known].groupby("Customer_ID")["Sales"].sum()
low_cut, high_cut=customer_spend.quantile([0.25, 0.75])

df["Customer_Total_Spend"]=df.groupby('Customer_ID')['Sales'].transform("sum").round(2)
df.loc[~known, "Customer_Total_Spend"] = np.nan 

df["Customer_Value_Segment"]=np.select(
    [~known,
    df["Customer_Total_Spend"]>=high_cut,
    df['Customer_Total_Spend']>=low_cut],
    ['Unknown','High Value','Medium Value'],
    default='Low Value'
)

```

```python
# Sales Category
q1,q3=df['Sales'].quantile([0.25,0.75])
df['Sales_Category']=pd.cut(df['Sales'],[-np.inf,q1,q3,np.inf],
                            labels=['Low Sales','Medium Sales','High Sales'])

# Profit Category
df['Profit_Category']=np.select(
    [df['Profit']<0,df['Profit_Margin']<10,df['Profit_Margin']<25],
    ['Loss','Low Profit','Medium Profit'],
    default='High Profit'
)

#Order Size
df['Order_Size']=pd.cut(df['Quantity'],[0,2,5,np.inf],labels=['Small','Medium','Large'])

#Customer Type
orders_per_customer=df.groupby('Customer_ID')['Order_ID'].transform("nunique")
df['Customer_Type']=np.select(
    [~known, orders_per_customer>1],
    ['Unknown','Repeat'],
    default='One-time'
)

# Product Performance
prod_profit=df.groupby('Product_Name')['Profit'].transform('sum')
p_low,p_high=df.groupby('Product_Name')['Profit'].sum().quantile([0.25,0.75])
df['Product_Performance']=np.select(
    [prod_profit>=p_high,prod_profit>=p_low],
    ['Top Performer','Average Performer'],
    default='Under Performer'
)

# Days to Ship
df['Days_To_Ship']=(df['Ship_Date']-df['Order_Date']).dt.days

# Discount Applied
df['Discount_Applied']=np.where(df['Discount']>0,'Yes','No')

# Day of Week
df['Day_Of_Week']=df['Order_Date'].dt.day_name()
```

```python
# Declining Products (year-over-year)
yearly=df.pivot_table(index='Product_Name',columns='Year',values='Sales',aggfunc='sum')

yearly['YoY_2023_%']=((yearly[2023]/yearly[2022]-1)*100).round(1)
yearly['YoY_2024_%']=((yearly[2024]/yearly[2023]-1)*100).round(1)

# Consistently declining: fell in BOTH 2023 and 2024
consistent_decline=yearly[(yearly[2023]<yearly[2022]) & (yearly[2024]<yearly[2023])]

# Recently declining: fell in 2024 vs 2023
recent_decline=yearly[(yearly[2024]<yearly[2023]) & (yearly[2023]>yearly[2022])].sort_values('YoY_2024_%')

print("Consistently declining (2022 > 2023 > 2024):",len(consistent_decline))
print(consistent_decline[[2022,2023,2024,'YoY_2023_%','YoY_2024_%']])
print("\nDeclined in 2024 vs 2023:",len(recent_decline))
print(recent_decline[[2022,2023,2024,'YoY_2024_%']])

# Chart: yearly sales of the declining products
fig,ax=plt.subplots(figsize=(9,5))
for name in consistent_decline.index:
    ax.plot(['2022','2023','2024'],consistent_decline.loc[name,[2022,2023,2024]].values,marker='o',label=name)
ax.set_title("Consistently Declining Products (Yearly Sales)",weight='bold')
ax.set_xlabel("Year")
ax.set_ylabel("Sales")
ax.legend()
save("18_declining_products")
```

```python
# Day 30 Data Model
# Dim Customer
cust = (df[df.Customer_ID != "Unknown"]
        .groupby("Customer_ID")
        .agg(Customer_Name=("Customer_Name","first"),
             Customer_Segment=("Segment","first"),
             Customer_Region=("Region", lambda s: s.mode().iat[0]),
             Customer_Value_Segment=("Customer_Value_Segment","first"),
             Customer_Type=("Customer_Type","first"))
        .reset_index())


unknown = pd.DataFrame([{"Customer_ID":"Unknown","Customer_Name":"Unknown Customer",
        "Customer_Segment":"Unknown","Customer_Region":"Unknown",
        "Customer_Value_Segment":"Unknown","Customer_Type":"Unknown"}])
dim_customer = pd.concat([cust,unknown],ignore_index=True)
```

```python
# Dim Product
dim_product = (df[["Product_ID","Product_Name","Category","Unit_Price","Product_Performance"]]
               .drop_duplicates("Product_ID").sort_values("Product_ID").reset_index(drop=True))
```

```python
#Dim Date
dates = pd.date_range(df.Order_Date.min().replace(month=1, day=1),
                      df.Order_Date.max().replace(month=12, day=31), freq="D")
dim_date = pd.DataFrame({"Date": dates})
dim_date["Year"]         = dim_date.Date.dt.year
dim_date["Month_Number"] = dim_date.Date.dt.month
dim_date["Month"]        = dim_date.Date.dt.month_name()
dim_date["Month_Short"]  = dim_date.Date.dt.strftime("%b")
dim_date["Quarter"]      = "Q" + dim_date.Date.dt.quarter.astype(str)
dim_date["Year_Month"]   = dim_date.Date.dt.strftime("%Y-%m")
dim_date["Day_Of_Week"]  = dim_date.Date.dt.day_name()
dim_date["Is_Weekend"]   = np.where(dim_date.Date.dt.dayofweek >= 5, "Yes","No")
```

```python
# Fact Sales
fact = df[["Order_ID","Customer_ID","Product_ID","Order_Date","Ship_Date",
           "Ship_Mode","Region","City","Quantity","Unit_Price","Discount",
           "Sales","Profit"]].copy()
fact = fact.rename(columns={"Region":"Ship_Region","City":"Ship_City"})
fact.insert(0, "Sales_Key", np.arange(1, len(fact)+1))
```

```python
assert dim_customer.Customer_ID.is_unique          # PK unique
assert dim_product.Product_ID.is_unique
assert dim_date.Date.is_unique
assert fact.Customer_ID.isin(dim_customer.Customer_ID).all()   # every FK has a parent
assert fact.Product_ID.isin(dim_product.Product_ID).all()
assert fact.Order_Date.isin(dim_date.Date).all()
assert not fact.duplicated(["Order_ID","Product_ID"]).any()    # grain check
assert np.isclose(fact.Sales.sum(), df.Sales.sum())            # nothing lost

for name, t in {"FactSales":fact,"DimCustomer":dim_customer,
                "DimProduct":dim_product,"DimDate":dim_date}.items():
    t.to_csv(f"{MODEL_DIR}/{name}.csv", index=False)
```
